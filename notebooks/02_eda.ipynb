# AeroPredict AI - 02 Exploratory Data Analysis
We investigate the sensor distributions, correlation with time/RUL, and identify which sensors contain the most useful degradation signals.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from src.data_loader import load_cmapss_data
from src.rul import calculate_rul

df = load_cmapss_data('train', 'FD001')
df = calculate_rul(df)

engine_1 = df[df['unit_number'] == 1]
plt.figure(figsize=(10, 4))
plt.plot(engine_1['time_cycles'], engine_1['sensor_2'])
plt.title('Sensor 2 Degradation for Engine 1')
plt.xlabel('Time Cycle')
plt.ylabel('Sensor 2 Value')
plt.show()
```
